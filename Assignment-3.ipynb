### Question: Count Vowels in a String

**Description:** Write a function named count_vowels that takes a string as an argument and returns the count of vowels (a, e, i, o, u) in the string. 

The function should be case-insensitive, counting both uppercase and lowercase vowels.


```python
'''
# Example Input:
text = "Hello World"
# Expected Output:
3
'''
```

```python
def count_vowels(text):
    return sum(map(lambda char: char.lower() in "aeiou", text))
print(count_vowels("Hello World"))
```

```python
count_vowels = lambda text: sum(1 for char in text.lower() if char in "aeiou")
print(count_vowels("Omar mohamed"))
```

### Question: Find the Maximum Value in a List Using a Function
Description: Write a function named find_max that takes a list of integers as an argument and returns the maximum value in the list. 

Do not use Python’s built-in max() function. The function should handle cases where the list is empty by returning None.


```python
'''
# Example Input:
numbers = [3, 5, 2, 8, 1]
# Expected Output:
8
'''
```

```python
def find_max(numbers):
    if not numbers:
        return None
    return sorted(numbers)[-1]
print(find_max([4, 9, 2, 7, 5]))
print(find_max([]))
```

```python
find_max = lambda numbers: None if not numbers else sorted(numbers)[-1]

print(find_max([4, 9, 2, 7, 5]))
print(find_max([]))
```
